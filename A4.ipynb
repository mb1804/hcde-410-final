# Motivation & problem 

Over the past few years, I have seen an increase in social media posts and comments that may stem from manosphere communities, whether the content uses their lingo or focuses on topics like improving one's looks or discussions on dating and relationships. These communities often spout toxic ideas and rhetoric. From this analysis, I hope to learn whether events such as presidential elections influence how toxic different manosphere communities become over the years.
