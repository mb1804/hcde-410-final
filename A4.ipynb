# Overview & research questions/hypotheses

In this project, I aim to determine whether different manosphere communities show changes in mean toxicity scores before, during, and after political and social events; compare these scores across platforms (subreddits and forums) and ideological categories; and see how they differ. 

For this project, I aim to explore the following research questions: 
How do major political and social events correlate with changes in mean toxicity scores in manosphere communities?
How do mean toxicity scores differ between similar ideologies on Reddit and forums during the events? 

My hypothesis is that mean toxicity scores will change significantly during events, with the change being greater on forums than on Reddit.

# Methodology
The analysis will use data from https://doi.org/10.5281/zenodo.4007913. I will merge the forum and subreddit data with the Perspective API annotations into a single dataset. To examine major political events such as presidential elections, I will filter the data starting in 2012, as this is not only a presidential election year but also includes years when the forums and accompanying subreddits have data. This will be the starting point and will run through the end of the dataset in 2018. I will then select a few events during that timeframe, such as the 2012 election, the 2016 election, and the #MeToo movement, to calculate mean toxicity scores 30 days before the event, during the event, and 30 days after. I will compare the scores across platforms (matched subreddits and forums, e.g., MGTOW Forum and MGTOW subreddits) and across ideologies. I will present these findings with line graphs showing changes in mean toxicity scores over time, as well as bar charts showing mean changes in toxicity for each event.
